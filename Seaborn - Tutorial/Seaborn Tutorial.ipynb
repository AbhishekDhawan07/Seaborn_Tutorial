# Line Plot

```python
import seaborn as sns
```

```python
# Creating sample dataset
data = {
    "Age": [22, 25, 23, 30, 28, 26, 24, 27],
    "Salary": [30000, 50000, 35000, 70000, 65000, 48000, 40000, 52000],
    "Department": ["IT", "HR", "IT", "Finance", "Finance", "HR", "IT", "HR"]
}
```

```python
sns.lineplot(x=data["Age"], y=data["Salary"]) # Shows trend between Age and Salary.
```

## Bar Plot

```python
sns.barplot(x=data["Department"], y=data["Salary"]) # Shows average salary per department.
```

## Histogram

```python
sns.histplot(data["Salary"], bins=5) # Shows salary distribution.
```

## Scatter Plot

```python
sns.scatterplot(x=data["Age"], y=data["Salary"]) # Shows relationship between Age and Salary.
```

## Box Plot

```python
sns.boxplot(x=data["Department"], y=data["Salary"]) # Shows salary spread in each department.
```

## Violin Plot

```python
sns.violinplot(x=data["Department"], y=data["Age"]) # Shows age distribution by department.
```

## Count Plot

```python
sns.countplot(x=data["Department"]) # Shows number of employees per department.
```
